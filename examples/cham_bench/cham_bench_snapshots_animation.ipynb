```python
import os
import numpy as np
import pandas as pd
from coregiver import composition_tracker_v1 as ctv1
```

```python
# cham_bench loop, uniform
n_runs = 50 # This variable is not used in the loop, you might want to remove it
cut_times = np.logspace(6, np.log10(299) + 6, 20)
for cut in cut_times:   
    for i in range(n_runs):
        collision_file = f"/Users/honeyeah/Codes/data/cham_bench/cham_bench_{i+1}/animation/filtered_collision_report_{int(cut/1e6)}Myrs.txt"
        comp_input_file = "/Users/honeyeah/Codes/coregiver/examples/cham_bench/uniform_33.txt"

        ejec_file = f"/Users/honeyeah/Codes/data/cham_bench/cham_bench_{i+1}/ejections.txt"

        if os.path.exists(ejec_file):
            print('ejections file found.', i)
        else:
            ejec_file = "/Users/honeyeah/Codes/planetgarten/body_composition_tracker/test_input/ejections_empty.txt"
            print("ejections file not found, using an empty file.", i)

        comp_output_file = f"/Users/honeyeah/Codes/data/cham_bench/cham_bench_{i+1}/animation/uniform_33_{int(cut/1e6)}Myrs.txt"

        final_compositions = ctv1.track_composition(collision_file, comp_input_file, ejec_file)
        ctv1.write_output(final_compositions, comp_output_file)
        print(f"Processed cham_bench run {i+1}")
```
