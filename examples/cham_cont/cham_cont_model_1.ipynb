# Step functions

```python
# cham_cont loop, for the step functions
cmf_grid = pd.read_csv('/Users/honeyeah/Codes/coregiver/examples/cham_cont/cham_cont_cmf_grid.csv')
bounds = cmf_grid['bounds']
percentages = cmf_grid['percentages']
n_runs = 60 # This variable is not used in the loop, you might want to remove it
for i in range(n_runs):
    for _, row in cmf_grid.iterrows():
        bound = row['bounds']
        perc = row['percentages']

        collision_file = f"/Users/honeyeah/Codes/data/cham_cont/cham_cont_{i+1}/collision_report.txt"
        comp_input_file = f"step_functions/cham_cont_step_b{bound}_p{perc}.txt"

        ejec_file = f"/Users/honeyeah/Codes/data/cham_cont/cham_cont_{i+1}/ejections.txt"

        if os.path.exists(ejec_file):
            print('ejections file found.', i)
        else:
            ejec_file = "/Users/honeyeah/Codes/planetgarten/body_composition_tracker/test_input/ejections_empty.txt"
            print("ejections file not found, using an empty file.", i)

        comp_output_file = f"/Users/honeyeah/Codes/data/cham_cont/cham_cont_{i+1}/cham_cont_output_step_b{bound}_p{perc}.txt"
        try: 
            final_compositions = ctv1.track_composition(collision_file, comp_input_file, ejec_file)
            ctv1.write_output(final_compositions, comp_output_file)
            print(f"Processed cham_cont run {i+1}")
        except Exception as e:
            print(f"Skipped Cham cont run {i+1} due to error: {e}")
            continue
```
