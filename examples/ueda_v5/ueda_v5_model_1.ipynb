# Uniform, Earth CMF

```python
# ueda_v5 loop, uniform
n_runs = 60 # This variable is not used in the loop, you might want to remove it
for i in range(0,n_runs):
    collision_file = f"/Users/honeyeah/Codes/data/ueda_v5/ueda_v5_{i+1}/collision_report.txt"
    comp_input_file = "/Users/honeyeah/Codes/coregiver/examples/ueda_v5/uniform_33.txt"

    ejec_file = f"/Users/honeyeah/Codes/data/ueda_v5/ueda_v5_{i+1}/ejections.txt"

    if os.path.exists(ejec_file):
        print('ejections file found.', i)
    else:
        ejec_file = "/Users/honeyeah/Codes/planetgarten/body_composition_tracker/test_input/ejections_empty.txt"
        print("ejections file not found, using an empty file.", i)

    comp_output_file = f"/Users/honeyeah/Codes/data/ueda_v5/ueda_v5_{i+1}/uniform_33.txt"
    try:
        final_compositions = ctv1.track_composition(collision_file, comp_input_file, ejec_file)
        ctv1.write_output(final_compositions, comp_output_file)
        print(f"Processed Ueda V3 run {i+1}")
    except Exception as e:
        print(f"Error occured in run {i}, skipping")
        continue
```

# Step functions

```python
# ueda_v5 loop, for the step functions
cmf_grid = pd.read_csv('/Users/honeyeah/Codes/coregiver/examples/ueda_v5/ueda_v5_cmf_grid.csv')
bounds = cmf_grid['bounds']
percentages = cmf_grid['percentages']
n_runs = 60 # This variable is not used in the loop, you might want to remove it
for i in range(0,n_runs):
    for _, row in cmf_grid.iterrows():
        bound = row['bounds']
        perc = row['percentages']

        collision_file = f"/Users/honeyeah/Codes/data/ueda_v5/ueda_v5_{i+1}/collision_report.txt"
        comp_input_file = f"step_functions/ueda_v5_step_b{bound}_p{perc}.txt"

        ejec_file = f"/Users/honeyeah/Codes/data/ueda_v5/ueda_v5_{i+1}/ejections.txt"

        if os.path.exists(ejec_file):
            print('ejections file found.', i)
        else:
            ejec_file = "/Users/honeyeah/Codes/planetgarten/body_composition_tracker/test_input/ejections_empty.txt"
            print("ejections file not found, using an empty file.", i)

        comp_output_file = f"/Users/honeyeah/Codes/data/ueda_v5/ueda_v5_{i+1}/uv5_output_step_b{bound}_p{perc}.txt"

        final_compositions = ctv1.track_composition(collision_file, comp_input_file, ejec_file)
        ctv1.write_output(final_compositions, comp_output_file)
        print(f"Processed ueda_v5 run {i+1}")
```
